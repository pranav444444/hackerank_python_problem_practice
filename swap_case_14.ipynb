```python
name = 'JiGnA'
result=" "
for i in range(len(name)):
  if name[i].isupper():
    result+=(name[i].lower())
  else:
    result+=(name[i].upper())
print(result)
```

```python
sentence="Pranav Is My Best Friend"
swapped_sentence=""

for i in range(len(sentence)):
  if sentence[i].isupper():
    swapped_sentence+=sentence[i].lower()
  else:
    swapped_sentence+=sentence[i].upper()
print(swapped_sentence)
```
