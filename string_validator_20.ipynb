```python

# My solution i did it using OOPS
class String_Validator:
    def __init__(self,s):
        self.s=s
        
    def alphanumeric(self):
        for x in range(len(self.s)):
            if self.s[x].isalnum():
                return True
        return False
            
    def alphabetical(self):
        for x in range(len(self.s)):
            if self.s[x].isalpha():
                return True
        return False
    def digits(self):
        for x in range(len(self.s)):
            if self.s[x].isdigit():
                return True
        return False
            
    def lowercase(self):
        for x in range(len(self.s)):
            if self.s[x].islower():
                return True
        return False
    def uppercase(self):
        for x in range(len(self.s)):
            if self.s[x].isupper():
                return True
        return False



if __name__ == '__main__':
    s = input()
    
    new_string=String_Validator(s)
    print(new_string.alphanumeric())
    print(new_string.alphabetical())
    print(new_string.digits())
    print(new_string.lowercase())
    print(new_string.uppercase())
```

```python
# Mentor's approach using any function

s=str(input())
```

```python
print(any(item.isalnum() for item in s))
print(any(item.isalpha() for item in s))
print(any(item.isdigit() for item in s))
print(any(item.islower() for item in s))
print(any(item.isupper() for item in s))
```
