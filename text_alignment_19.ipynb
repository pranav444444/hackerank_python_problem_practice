ljust stands for left justified

```python
'H'.ljust(7)
```

rjust stands for right Justified

```python
'H'.rjust(7)
```

center stands for center justified

```python
'H'.center(7)
```

```python
thickness = int(input()) #This must be an odd number
c = 'H'
```

```python
thickness
```

```python
#Top Cone
for i in range(thickness):
    print((c*i).rjust(thickness-1)+c+(c*i).ljust(thickness-1))
```

```python
#Top Pillars
for i in range(thickness+1):
    print((c*thickness).center(thickness*2)+(c*thickness).center(thickness*6))
```

```python
#Top Cone
for i in range(thickness):
    print((c*i).rjust(thickness-1)+c+(c*i).ljust(thickness-1))
#Top Pillars
for i in range(thickness+1):
    print((c*thickness).center(thickness*2)+(c*thickness).center(thickness*6))
```

```python
#Middle Belt
for i in range((thickness+1)//2):
    print((c*thickness*5).center(thickness*6)) 
```

```python
#Bottom Cone
for i in range(thickness):
    print(((c*(thickness-i-1)).rjust(thickness)+c+(c*(thickness-i-1)).ljust(thickness)).rjust(thickness*6)) #the last rjust is used to justify the whole bottom cone towards right side .. if we remove it then it will print below at left side
```
